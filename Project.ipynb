```python
import pandas as pd
import numpy as np
# Load your original Excel  file
df=pd.read_excel("/content/Proteins (AutoRecovered) (1).xlsx")
# Display first 10 rows
display(df.head(10))
# Check dataset dimensions
print("Dataset shape:",df.shape)
# Check column names
print("\nColumn names:")
print(df.columns.tolist())
# Check missing values
print("\nMissing values:")
print(df.isnull().sum())
# Check data types
print("\nData types:")
print(df.dtypes)
```

```python
print("Number of proteins:",len(df))
print("Duplicate rows:",df.duplicated().sum())
print("Dataset information:")
df.info()
```

```python
# Check duplicate proteins
protein_col=df.columns[0]
print("Total entries:",len(df))
print("Unique protein entries:",df[protein_col].nunique())
print("Duplicate protein entries:",df[protein_col].duplicated().sum())
duplicates=df[df[protein_col].duplicated(keep=False)]
display(duplicates.sort_values(by=protein_col))
```
